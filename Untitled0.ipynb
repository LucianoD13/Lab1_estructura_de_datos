```python
t = int(input("Ingrese casos:"))

for _ in range(t):
    n = int(input("Ingrese largo de la cadena:"))
    s = input("Ingrese cadena:")

    usados = set()
    total = 0

    for i in range(n):
        usados.add(s[i])
        print("Caracteres distintos hasta ahora:",usados)
        total += len(usados)
        print("Se han sumado:",len(usados), "total actualizado =", total)

    print("La cantidad de palabras formables y distintas es:",total)
```
